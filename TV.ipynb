```python
df["Channel_Name"] = df["Channel_Name"].str.strip().str.title()
print(df["Channel_Name"].unique())
```

```python
numerical_columns = df.select_dtypes(include="number").columns
for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())
categorical_columns = df.select_dtypes(include="object").columns
for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
df = df.drop_duplicates()
print("Duplicate records:", df.duplicated().sum())
```

```python
df.loc[
    (df["Avg_Rating"] < 0) | (df["Avg_Rating"] > 5),
    "Avg_Rating"
] = None

df["Avg_Rating"] = df["Avg_Rating"].fillna(
    df["Avg_Rating"].median()
)
```

```python
df.loc[
    df["Subscription_Fee"] < 0,
    "Subscription_Fee"
] = None

df["Subscription_Fee"] = df["Subscription_Fee"].fillna(
    df["Subscription_Fee"].median()
)
```

```python
df.loc[
    (df["Years_Active"] < 0) |
    (df["Years_Active"] > 100),
    "Years_Active"
] = None

df["Years_Active"] = df["Years_Active"].fillna(
    df["Years_Active"].median()
)
```

```python
df["Avg_Rating"] = pd.to_numeric(
    df["Avg_Rating"], errors="coerce"
)

df["Subscription_Fee"] = pd.to_numeric(
    df["Subscription_Fee"], errors="coerce"
)

df["Years_Active"] = pd.to_numeric(
    df["Years_Active"], errors="coerce"
)

df["Avg_Viewers"] = pd.to_numeric(
    df["Avg_Viewers"], errors="coerce"
)
```

```python
import numpy as np

df = df.replace(
    ["NA", "N/A", "na", "n/a", "unknown", "Unknown", ""],
    np.nan
)
```

```python

for column in df.select_dtypes(include="number").columns:
    df[column] = df[column].fillna(df[column].median())
for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].fillna(df[column].mode()[0]);
```
